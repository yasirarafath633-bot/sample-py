```python
x=int(input("Enter no.of.rows :"))
y=int(input("Enter no.of.columns :"))
symbol=input("Give any symbol : (!/@/#/$/%/^/&) :")


a=1
while a<=x:
    b=1
    while b<=y:
        print(symbol, end="")
        b+=1
    print()
    a+=1
```

```python
n = int(input("Enter the number of rows: "))
m = int(input("Enter the number of columns: "))

for i in range(n):
    for j in range(m):
        print("*", end='')
        print()
```
