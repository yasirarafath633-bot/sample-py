### Print numbers from 1 to 10 using a for loop ###

```python
for i in range (1,11):
    print(i)
```

### Print numbers from 10 to 1 using a while loop. ###

```python
i = 10 
while i >= 1:
    print(i)
    i -= 1  
```

###  Write a program to print the multiplication table of a given number. ###

```python
num = int(input("Enter a number: "))

for i in range (1, 11):
    print (num, "x",i, "=", num *i)
```

### Write a program to find the sum of first N natural numbers using a loop. ###

```python
n = int(input("Enter N: "))

sum = 0

for i in range(1, n + 1):
    sum = sum + i

print("Sum =", sum)
```

###  Write a function add_numbers(a, b) that returns the sum of two numbers. ###

```python
def add_numbers(a, b):
    return a + b

result = add_numbers(600,400)
print("Sum =", result)
```

### Write a function is_even(n) that returns True if the number is even, otherwise False. ###

```python
def is_even(n):
    return n % 2 == 0

print(is_even(16))
print(is_even(19))
```

###  Write a function square(n) that returns the square of a given number. ###

```python
def square(n):
    return n * n

print(square(7))
```

### Write a function max_of_two(a, b) that returns the larger of two numbers #####

```python
def max_of_two(a, b):
    if a > b:
        return a
    else:
        return b

print(max_of_two(20, 30))
```
