### Symbol Question ###         

```python
a=int(input("enter a number :"))
symbol=input("enter a symbol :")

for x in range(a):
    for y in range(0, x+1):
        print(symbol, end='')
        print()
```

# Question Patterns ### 

```python
symbol=input("Give any Symbol:(!/@/#/$/%^&/*) :")
row=int(input("Enter no.of.rows :"))


for i in range(row):
    for j in range(0,i+1):
        print(symbol,end='')
        print()
```
